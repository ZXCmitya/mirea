```python
def qZ(x, y):
  return (x - 3 * y + 1) / (3 * x ** 2 + 3 * y ** 2 + 1)

def qSumZ(Z):
  return sum(Z)

def exchangeScheme(oldX, oldY, sortedId):
  newX = [0 for i in range(4)]
  newY = [0 for i in range(4)]
  newX[2] = oldX[sortedId[2]]
  newX[3] = oldX[sortedId[2]]

  newX[0] = oldX[sortedId[0]]

  newX[1] = oldX[sortedId[1]]

  newY[0] = oldY[sortedId[2]]
  newY[1] = oldY[sortedId[2]]
  newY[2] = oldY[sortedId[0]]

  newY[3] = oldY[sortedId[1]]

  return newX, newY

def sorting(Z):
  sortedId = sorted(range(len(Z)), key=lambda k: Z[k])
  return sortedId

def Step(x, y, z):
  _, minId = min((value, id) for (id, value) in enumerate(z))
  x = x[:]
  y = y[:]
  z = z[:]

  x.pop(minId)
  y.pop(minId)
  z.pop(minId)

  return x, y, z

def doSteps(X, Y, count = 4):
  res = []

  for j in range(count):
    arrZ = [qZ(x, Y[i]) for i, x in enumerate(X)]

    X, Y, Z = Step(X, Y, arrZ)
    X, Y = exchangeScheme(X, Y, sorting(Z))
    res.append([X, Y, qSumZ(arrZ), arrZ])

  return X, Y, res

ans = doSteps(X, Y)

for i in range(len(ans[2])):
  print(f'max_{i + 1}_step: {ans[2][i][2]}')
  qualityArrZ = []
for i in range(len(ans[2])):
  qualityArrZ += ans [2][1][3]
print(f'max Z: {max(qualityArrZ)}')
```
