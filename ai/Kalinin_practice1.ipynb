```python
from numpy import exp, cos, sin, sqrt, log
x = [i for i in range(1, 11)]
f = []

for i in x:
    func = (sqrt(1 + exp(sqrt(i)) + cos(i ** 2))) / (abs(1 - sin(i) ** 3)) + log(abs(2 * i))
    f.append(func)

plt.plot(x, f)
plt.scatter(x[:5], f[:5])
```

```python
import matplotlib.pyplot as plt
import yfinance as yf

tickers = ['AAPL', 'MSFT', 'GOOGL']
data = yf.download(tickers, start='2021-01-01', end='2021-12-31')['Close']

fig, axes = plt.subplots(3, 1, figsize=(8, 10))

for i, ticker in enumerate(tickers):
  axes[i].plot(data.index, data[ticker])
  axes[i].set_title(f'Акции {ticker}')
  axes[i].set_ylabel('Цена ($)')
  axes[i].grid(True)

plt.xlabel('Дата')
plt.tight_layout()
plt.show()

print('Рост за 2021 год:')
for ticker in tickers:
  start_price = data[ticker].iloc[0]
  end_price = data[ticker].iloc[-1]
  growth = ((end_price - start_price) / start_price) * 100
  print(f'{ticker}: {growth:.1f}%')
```

```python
print('Формат записи: A + B (или номер функции: 1, 2, 3, 4)')
print('1: вычисление значения функции f(x, y) = exp(x + y)')
print('2: вычисление значения функции f(x, y) = sin(x + y)')
print('3: вычисление значения функции f(x, y) = cos(x + y)')
print('4: вычисление значения функции f(x, y) = x^y')

req = input().split()

try:
    a = int(req[0])
    func = req[1]
    b = int(req[2])

    if func == '+':
        print(a + b)
    elif func == '-':
        print(a - b)
    elif func == '*':
        print(a * b)
    elif func == '/':
        if b != 0:
            print(a / b)
        else:
            print("Ошибка: деление на ноль!")
    else:
        print("Неизвестная операция!")

except IndexError:
    try:
        single_num = int(req[0])
        if single_num in (1, 2, 3, 4):
            x = int(input('x = '))
            y = int(input('y = '))

            if single_num == 1:
                print(exp(x + y))
            elif single_num == 2:
                print(sin(x + y))
            elif single_num == 3:
                print(cos(x + y))
            elif single_num == 4:
                print(x ** y)
        else:
            print("Ошибка: выберите пункт от 1 до 4!")

    except IndexError:
        print("Ошибка: вы ничего не ввели!")
    except ValueError:
        print("Ошибка: введены некорректные данные (не число)!")

except ValueError:
    print("Ошибка: введены некорректные данные (не число)!")
```
