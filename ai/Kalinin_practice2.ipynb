```python
url = "https://raw.githubusercontent.com/akmand/datasets/master/iris.csv"
df = pd.read_csv(url)

min_val = df['sepal_length_cm'].min()
max_val = df['sepal_length_cm'].max()
df['sepal_length_norm'] = (df['sepal_length_cm'] - min_val) / (max_val - min_val)

mean_val = df['sepal_width_cm'].mean()
std_val = df['sepal_width_cm'].std()
df['sepal_width_z'] = (df['sepal_width_cm'] - mean_val) / std_val
```
