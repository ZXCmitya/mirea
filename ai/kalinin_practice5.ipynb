```python
from math import sin, cos, tan, asin, acos, atan, radians

class Trigonometry:

    def SolveCos(self, arg):
        return cos(arg)

    def SolveSin(self, arg):
        return sin(arg)

    def SolveTan(self, arg):
        return tan(arg)

    def SolveAsin(self, arg):
        return asin(arg)

    def SolveAcos(self, arg):
        return acos(arg)

    def SolveAtan(self, arg):
        return atan(arg)

    def Deg2rad(self, arg):
        return radians(arg)

trig_xmpl = Trigonometry()
print(trig_xmpl.SolveSin(0))
print(trig_xmpl.SolveCos(0))
print(trig_xmpl.SolveTan(0))
print(trig_xmpl.SolveAsin(1))
print(trig_xmpl.SolveAcos(1))
print(trig_xmpl.SolveAtan(1))
print(trig_xmpl.Deg2rad(180))
```

```python
class Tree:
    def __init__(self, data):
        self.left = None
        self.right = None
        self.data = data

    def PrintTree(self):
        if self.left:
            self.left.PrintTree()
        print(self.data)
        if self.right:
            self.right.PrintTree()

    def Insert(self, data):
        if data < self.data:
            if self.left is None:
                self.left = Tree(data)
            else:
                self.left.Insert(data)
        elif data > self.data:
            if self.right is None:
                self.right = Tree(data)
            else:
                self.right.Insert(data)
        else:
            print('Value already exists')

root_val = input('Value (или stop для выхода): ')
if root_val.lower() != 'stop':
    tree = Tree(int(root_val))

    while True:
        val = input('Value: ')
        if val.lower() == 'stop':
            break
        tree.Insert(int(val))

    print("\nПолное дерево:")
    tree.PrintTree()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import classification_report, confusion_matrix, mean_squared_error, mean_absolute_error

x = np.array([[-1, -1], [-2, -1], [-3, -2], [1, 1], [2, 1], [3, 2]])
y = [0, 0, 0, 1, 1, 1]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)

model = DecisionTreeClassifier()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)

print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))

```

```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor

url = 'https://raw.githubusercontent.com/aniruddhachoudhury/Red-Wine-Quality/master/winequality-red.csv'

data = pd.read_csv(url)
data

x = data.iloc[:, :-1]
y = data.iloc[:, -1]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)

model = DecisionTreeRegressor()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)

# plot_tree(model)

df=pd.DataFrame({'Actual':y_test, 'Predicted':y_pred})

plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, color='blue', alpha=0.5)

plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color='red', lw=2)

plt.xlabel('Реальные значения (Actual)')
plt.ylabel('Предсказанные значения (Predicted)')
plt.title('Сравнение реальных и предсказанных значений')
plt.grid(True)
plt.show()

print('Mean Squared Error:', mean_squared_error(y_test, y_pred))
print('Mean Absolute Error:', mean_absolute_error(y_test, y_pred))
```
